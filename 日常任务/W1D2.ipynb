```python
# 练习1-2：输出学生成绩
students = [
    {"学号": "001", "姓名": "张三", "班级": "软件工程232", "高数": 85, "英语": 92, "Java编程": 78},
    {"学号": "002", "姓名": "李四", "班级": "软件工程232", "高数": 59, "英语": 88, "Java编程": 95},
    {"学号": "003", "姓名": "王五", "班级": "软件工程232", "高数": 73, "英语": 64, "Java编程": 82},
    {"学号": "004", "姓名": "赵六", "班级": "软件工程232", "高数": 91, "英语": 97, "Java编程": 89},
    {"学号": "005", "姓名": "孙七", "班级": "软件工程232", "高数": 55, "英语": 60, "Java编程": 58},
]

# 初始化
mathmax = 0
enmax = 0
javamax = 0
allmax = 0

math_first_name = ""
en_first_name = ""
java_first_name = ""
all_first_name = ""

# 挂科名单
fail_math = []      # 高数挂科
fail_english = []   # 英语挂科
fail_java = []      # Java挂科
fail_all = []       # 所有挂科学生（含科目）

# 遍历学生
for item in students:
    # 单科最高分及姓名
    if item["高数"] > mathmax:
        mathmax = item["高数"]
        math_first_name = item["姓名"]
    if item["英语"] > enmax:
        enmax = item["英语"]
        en_first_name = item["姓名"]
    if item["Java编程"] > javamax:
        javamax = item["Java编程"]
        java_first_name = item["姓名"]

    # 总分最高分及姓名
    total = item["高数"] + item["英语"] + item["Java编程"]
    if total > allmax:
        allmax = total
        all_first_name = item["姓名"]

    # 挂科判断（低于60分）
    fail_courses = []
    if item["高数"] < 60:
        fail_math.append(item["姓名"])
        fail_courses.append("高数")
    if item["英语"] < 60:
        fail_english.append(item["姓名"])
        fail_courses.append("英语")
    if item["Java编程"] < 60:
        fail_java.append(item["姓名"])
        fail_courses.append("Java编程")

    if fail_courses:
        fail_all.append({"姓名": item["姓名"], "挂科科目": fail_courses})

# 输出单科和总分最高分名单
print(f"高数最高分：{mathmax}（{math_first_name}）")
print(f"英语最高分：{enmax}（{en_first_name}）")
print(f"Java编程最高分：{javamax}（{java_first_name}）")
print(f"总分最高分：{allmax}（{all_first_name}）")
print()

# 输出各科目挂科名单
print(f"高数挂科：{fail_math if fail_math else '无'}")
print(f"英语挂科：{fail_english if fail_english else '无'}")
print(f"Java编程挂科：{fail_java if fail_java else '无'}")
print()
```

```python
# 练习2-1：用户登录
SAVED_USERNAME = "admin"
SAVED_PASSWORD = "123456"
MAX_ATTEMPTS = 3

attempts = 0

print("欢迎登录系统")
print(f"您有 {MAX_ATTEMPTS} 次尝试机会\n")

while attempts < MAX_ATTEMPTS:
    username = input("请输入用户名：")
    password = input("请输入密码：")

    if username == SAVED_USERNAME and password == SAVED_PASSWORD:
        print("登录成功！欢迎回来")
        break
    else:
        attempts += 1
        remaining = MAX_ATTEMPTS - attempts
        if remaining > 0:
            print(f"用户名或密码错误，还剩 {remaining} 次机会\n")
        else:
            print("尝试次数已达上限，账户已锁定")
```

```python
# 练习2-2-1 十进制转二进制
def decimal_to_binary(n):
    """将十进制数转换为二进制字符串"""
    if n == 0:
        return "0"

    # 处理负数
    negative = False
    if n < 0:
        negative = True
        n = -n

    binary = ""
    while n > 0:
        remainder = n % 2      # 余数（0或1）
        binary = str(remainder) + binary  # 拼接在前面
        n = n // 2             # 整除

    if negative:
        binary = "-" + binary

    return binary

num = int(input("请输入一个十进制数："))
result = decimal_to_binary(num)
print(f"二进制：{result}")
```

```python
# 练习2-2-2 十进制转十六进制
def decimal_to_hexadecimal(n):
    """将十进制数转换为十六进制字符串"""
    if n == 0:
        return "0"

    # 十六进制字符映射表
    hex_chars = "0123456789ABCDEF"

    # 处理负数
    negative = False
    if n < 0:
        negative = True
        n = -n

    hexadecimal = ""
    while n > 0:
        remainder = n % 16
        hexadecimal = hex_chars[remainder] + hexadecimal
        n = n // 16

    if negative:
        hexadecimal = "-" + hexadecimal

    return hexadecimal

# 测试
num = int(input("请输入一个十进制数："))
result = decimal_to_hexadecimal(num)
print(f"十六进制：{result}")
```

```python
# 练习 3-2 用户类与管理员类
class User:
    """用户类"""

    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    def describe_user(self):
        """打印用户信息"""
        print(f"用户信息：{self.first_name} {self.last_name}")

    def greet_user(self):
        """打印问候语"""
        print(f"你好，{self.first_name} {self.last_name}！欢迎回来。")


class Admin(User):
    """管理员类，继承自User"""

    def __init__(self, first_name, last_name):
        super().__init__(first_name, last_name)
        self.privileges = [
            "can add post",
            "can delete post",
            "can ban user",
            "can modify settings"
        ]

    def show_privileges(self):
        """显示管理员的权限"""
        print(f"管理员 {self.first_name} {self.last_name} 的权限：")
        for privilege in self.privileges:
            print(f"  - {privilege}")


if __name__ == '__main__':
    # 创建普通用户
    user1 = User("张三", "小明")
    user1.describe_user()
    user1.greet_user()

    print()

    # 创建管理员
    admin1 = Admin("李", "四")
    admin1.describe_user()
    admin1.greet_user()
    admin1.show_privileges()
```
